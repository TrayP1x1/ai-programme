**The mental model.** Git keeps your work in three places:
1. the **working directory**: the files you edit,
2. the **staging area**: what will go into the next commit (`git add` puts changes here),
3. the **history**: the commits.

Almost every command moves changes between these three. After every step below, run `git status` and `git log --oneline --graph --all`, and **predict the output before you look**.

#### Task 1: Build some history
Create `notes.txt` with one line, then `add` and `commit` it. Make two more commits, each one adding a line.

#### Task 2: See what changed
1. Edit `notes.txt`. Run `git diff`.
2. `git add notes.txt`, then run `git diff` again. It's empty now. Why?
3. Run `git diff --staged`.

#### Task 3: Undo before committing
1. `git restore --staged notes.txt` unstages the change. The edit stays in your file.
2. `git restore notes.txt` throws the edit away. **It's gone for good**, because Git never saw it.

#### Task 4: Fix the last commit
1. Commit with a typo in the message, then fix it with `git commit --amend -m "Better message"`.
2. Commit, then notice you forgot a file. Create `todo.txt`, `git add` it, then `git commit --amend --no-edit`.

Only amend commits you **haven't pushed yet**. Amending rewrites history, and GitHub will reject the push.

#### Task 5: Stash: "I need a clean state for a minute"
1. Edit `notes.txt`, but don't commit.
2. `git stash`. Then `git status` shows a clean working directory. `git stash list` shows your saved change.
3. `git stash pop` brings it back.

You'll need this when Git refuses to switch branches because you have uncommitted changes.

#### Task 6: Undo a commit, two ways
1. `git reset --soft HEAD~1` removes the last commit and keeps its changes staged. Recommit it.
2. `git revert HEAD` adds a **new** commit that undoes the last one. Compare the two logs.

Rule: use `reset` for commits that only exist on your machine, and `revert` for commits you've already pushed.

#### Task 7: Your first merge conflict
1. `git switch -c experiment`, change **line 1** of `notes.txt`, and commit.
2. `git switch main`, change **line 1** differently, and commit.
3. `git merge experiment`. Git stops with a conflict.
4. Open `notes.txt` and find the `<<<<<<<`, `=======` and `>>>>>>>` markers. Edit the file so it says what you want, delete the markers, then `git add notes.txt` and `git commit`.

Week 2 builds on this with pull requests.

#### Task 8: Nothing committed is ever really lost
1. `git reset --hard HEAD~2`. Two commits are gone from `git log`.
2. `git reflog` lists everywhere `HEAD` has been. Find the hash from before the reset.
3. `git reset --hard <that hash>`. Your commits are back.

#### Error messages you'll see, and what to do

| Message (shortened) | What it means | Fix |
|---|---|---|
| `nothing added to commit` | You forgot `git add` | `git add <file>`, then commit |
| `Your local changes ... would be overwritten` | Uncommitted edits block a switch, pull or merge | Commit them, or `git stash` |
| `rejected ... (fetch first)` / `non-fast-forward` | GitHub has commits you don't have | `git pull`, fix any conflict, then `git push` |
| `CONFLICT (content): Merge conflict in ...` | Two branches changed the same lines | Task 7 |
| `You are in 'detached HEAD' state` | You checked out a commit, not a branch | `git switch main` (or `git switch -c new-branch` to keep the work) |
| `Permission denied (publickey)` | Your SSH key isn't being used | `ssh -T git@github.com`, then check `git remote -v` uses `git@github.com:` |

### 2.1 Values and types
Every value in Python has a **type**, and the type decides what you can do with the value. Predict each output *before* you run the cell.

````python
age = 21
height = 1.82
name = "Tray"
is_student = True
print(type(age), type(height), type(name), type(is_student))

print(7 / 2)    # true division always gives a float
print(7 // 2)   # floor division
print(7 % 2)    # remainder: this is how you test even/odd
print(2 ** 10)  # power
print("abc" * 3, "abc" + "def")
````

### 2.2 Lists: ordered, changeable collections
Indexing starts at **0**. Negative indices count from the end. A slice `a[i:j]` includes `i` and **excludes** `j`.

````python
scores = [72, 85, 90, 64, 88]
print(scores[0], scores[-1])   # first, last
print(scores[1:3])             # items at index 1 and 2
print(len(scores))
scores.append(95)
print(scores)
print(sum(scores), max(scores), sorted(scores))
````

### 2.3 Loops and conditions
`for x in collection:` runs the indented block once for each item. Python uses **indentation** to decide what belongs inside the loop.

````python
for s in scores:
    if s >= 85:
        print(s, "high")
    elif s >= 70:
        print(s, "ok")
    else:
        print(s, "low")

for i in range(3):          # range(3) gives 0, 1, 2
    print("iteration", i)

for i, s in enumerate(scores):   # gives index and value together
    print(i, s)
````

### 2.4 Functions: `return` versus `print`
A function is a small machine: values go in as **arguments**, and one value comes out through **`return`**. `print` only *shows* something on screen; the caller can't use it. Nearly every function you write in this course should `return` its result.

````python
def mean(values):
    """Return the arithmetic mean of a non-empty list of numbers."""
    total = 0
    for v in values:
        total += v          # same as: total = total + v
    return total / len(values)

m = mean([1, 2, 3, 4])
print(m, m * 2)            # we can keep calculating with m because it was returned

def mean_printing(values):
    print(sum(values) / len(values))

result = mean_printing([1, 2, 3, 4])
print("result is", result)  # None: nothing was returned
````

### 3.2 Dictionaries: look up a value by its key
A dict maps **keys** to **values**. Looking up a key takes about the same time however big the dict is, which is why dicts are everywhere in AI code: vocabularies, word counts, configuration, and inverted indexes.

````python
counts = {}
for word in ["mri", "ct", "mri", "xray", "mri"]:
    counts[word] = counts.get(word, 0) + 1    # .get gives 0 if the key is missing
print(counts)
print(counts["mri"])
print("pet" in counts)

for word, n in counts.items():
    print(word, "->", n)

pairs = list(counts.items())
print(sorted(pairs, key=lambda item: item[1], reverse=True))   # sort by count
````

# Day 4: Reading errors and fixing bugs (2 h)

### 4.1 How to read a traceback
Read it from the **bottom up**. The last line gives the **error type** and **message**. The lines above show the chain of calls, ending at the line that failed.

````python
import traceback

def get_third(items):
    return items[3]          # bug: the third item has index 2

try:
    get_third([10, 20, 30])
except Exception:
    traceback.print_exc()
````

| Error | Usually means |
|---|---|
| `NameError` | A typo in a name, or using a variable before you created it |
| `TypeError` | The wrong kind of value, e.g. `"3" + 4`, or the wrong number of arguments |
| `IndexError` | A list index past the end (often off by one) |
| `KeyError` | A dict key that doesn't exist (use `.get` or check with `in`) |
| `AttributeError` | Calling a method the object doesn't have, e.g. `None.append` |
| `ZeroDivisionError` | Dividing by zero, often because of an empty list |
| `IndentationError` | Inconsistent indentation |

### 4.2 The debugging method (use it every time)
1. **Reproduce** the bug with the smallest input you can.
2. **Read** the traceback, or compare what you expected with what you got.
3. **Hypothesise**: "I think X happens because Y."
4. **Inspect**: add `print(...)` or use the debugger to test the hypothesis.
5. **Fix** one thing at a time.
6. **Lock it in** with a test, so the bug can't quietly come back.

#### Bug 4: `find_max`

````python
def find_max(nums):
    best = nums[0]
    for n in nums:
        if n > best:
            best = n
        return best
````

````python
assert find_max([3, 9, 2]) == 9
assert find_max([5]) == 5
print("✅ bug 4 fixed")
````
