### A1. Sum of even numbers
Write `total_even(nums)`, which returns the sum of the even numbers in a list. A number `n` is even when `n % 2 == 0`.

````python
def total_even(nums):
    return sum([number for number in nums if number % 2 == 0])
   
````

````python
assert total_even([1, 2, 3, 4]) == 6
assert total_even([]) == 0
assert total_even([1, 3, 5]) == 0
assert total_even([-2, 2, 7]) == 0
print("✅ A1 passed")
````

### A2. Letter frequencies
Write `char_frequency(text)`, which returns a dictionary mapping each **letter** to how many times it appears. Ignore case (so `"A"` and `"a"` count as the same letter) and skip anything that isn't a letter. `str.isalpha()` will help.

````python
from collections import defaultdict
def char_frequency(text):
    out = defaultdict(int)

    for char in text:
        if char.isalpha():
            c = char.lower()
            out[c] += 1

    return out 

    
````

````python
print(char_frequency("hello worLd!"))
````

````python
assert char_frequency("Aab!") == {"a": 2, "b": 1}
assert char_frequency("") == {}
assert char_frequency("1 2 3") == {}
assert char_frequency("Hello World") == {"h": 1, "e": 1, "l": 3, "o": 2, "w": 1, "r": 1, "d": 1}
print("✅ A2 passed")
````

### A4. A small class
Write a class `BankAccount` that:
- starts with `balance` equal to the `opening_balance` you pass in (default `0`),
- has `deposit(amount)`, which adds money,
- has `withdraw(amount)`, which subtracts money but raises `ValueError` if there isn't enough,
- keeps a list `history` of every successful transaction, as `("deposit", amount)` or `("withdraw", amount)`.

````python
class BankAccount:
    def __init__(self, opening_balance = 0):
        self.balance = opening_balance
        self.history = []

    def deposit(self, amount):
        if amount >  0: 
            self.balance += amount
            self.history.append(("deposit", amount))
        else:
            raise ValueError("Amount should be greater than 0")

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            self.history.append(("withdraw", amount))
        else:
            raise ValueError("There isnt enough")

    
    
````

````python
acc = BankAccount(10)
acc.deposit(5)
acc.withdraw(3)
assert acc.balance == 12
assert acc.history == [("deposit", 5), ("withdraw", 3)]
try:
    acc.withdraw(100)
    raise AssertionError("should not allow overdraft")
except ValueError:
    pass
assert acc.balance == 12 and len(acc.history) == 2
assert BankAccount().balance == 0
print("✅ A4 passed")
````
