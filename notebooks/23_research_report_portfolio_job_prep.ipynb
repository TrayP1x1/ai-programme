# Days 1–2: The research report (5 h)

**Structure** (use exactly these `#`/`##` headings; the checker looks for them):
1. `Abstract`: 150 words covering the problem, method, main results (with numbers and CIs), and the main limitation.
2. `Introduction`: why biomedical search and medical image classification matter; your two research questions; your contributions (3 bullets).
3. `Related work`: BM25 (Robertson and Zaragoza, 2009), BEIR (Thakur et al., 2021), Sentence-BERT (Reimers and Gurevych, 2019), MedMNIST (Yang et al., 2023), transfer learning for medical imaging (for example Raghu et al., 2019), calibration (Guo et al., 2017). One paragraph each at most. **Cite only papers you've actually opened.**
4. `Data`: summarise `DATASETS.md`.
5. `Methods`: systems, training details, the evaluation protocol (pointing to `PROTOCOL.md`), and the statistical tests.
6. `Results`: Table A and Table B, plus the figures. State the results; don't interpret yet.
7. `Discussion`: interpretation; hypotheses supported or not; error analysis; the robustness and calibration findings.
8. `Limitations`: be specific (judgement incompleteness, low resolution, no patient-level split info, synthetic shift, single dataset per task, compute limits).
9. `Ethics`: public data only, no patient data handled, **not clinically validated**, risks of automation bias if anything like this were deployed.
10. `Reproducibility`: the exact commands, the commit hash, hardware, and runtime.
11. `References`.

**Writing tips:** one idea per paragraph; numbers always come with uncertainty ("0.334 [0.301, 0.366]"); write "we did not observe a significant difference", not "there is no difference"; every figure is referenced in the text and has a caption that stands on its own.

Export to PDF with `pandoc report.md -o report.pdf` (optional), or keep the Markdown. GitHub renders it nicely.

# Day 3: The portfolio README (2 h)

Recruiters and hiring managers spend about a minute on a repo. Your README must answer, **in this order**:
1. **What is it?** (one sentence plus a figure or GIF from the demo),
2. **Key results** (a small table with CIs),
3. **Quick start** (≤ 5 commands, copy-pasteable),
4. **What's inside** (the repo structure, with one line per folder),
5. **Reproducing the paper** (`scripts/reproduce.sh`, expected runtime, GPU note),
6. **Data and licences** (link `DATASETS.md`), **Limitations**, the **not clinically validated** disclaimer,
7. **Tech used** (Python, NumPy, PyTorch, scikit-learn, FastAPI, Docker, GitHub Actions, SQLite) and the CI badge,
8. **Licence** for your code (MIT is the common choice) and how to cite the datasets.

Use these `##` headings: `Key results`, `Quick start`, `Repository structure`, `Reproducing`, `Data`, `Limitations`, `License`.
