# Day 1: Recursion (2.5 h)

A function is **recursive** when it calls itself on a *smaller* version of the problem. Every recursive function needs:
1. a **base case**: an input small enough to answer directly,
2. a **recursive case**: break the problem down, call yourself on the smaller part, and combine the results.

Each call waits on the **call stack** until the calls it made return. The demo below prints that stack.

````python
def factorial(n, depth=0):
    print("  " * depth + f"factorial({n})")
    if n <= 1:                      # base case
        return 1
    result = n * factorial(n - 1, depth + 1)   # recursive case
    print("  " * depth + f"-> {result}")
    return result

factorial(4)
````

**How to design a recursive function (the "leap of faith"):** *assume* the function already works for smaller inputs, and ask only "how do I build the answer for `n` from the answer for a smaller input?" Trees, nested lists, divide-and-conquer sorting, and backpropagation through a computation graph all have this recursive shape.
